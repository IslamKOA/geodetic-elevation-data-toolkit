```python
import sys
import subprocess
import importlib
import importlib.util

REQUIRED_PACKAGES = {
    "ipywidgets": "ipywidgets",
    "geopandas": "geopandas",
    "rasterio": "rasterio",
    "pyproj": "pyproj",
    "matplotlib": "matplotlib",
    "numpy": "numpy",
    "pandas": "pandas",
    "shapely": "shapely",
    "affine": "affine",
    "scipy": "scipy",
}

missing = [
    package
    for module, package in REQUIRED_PACKAGES.items()
    if importlib.util.find_spec(module) is None
]

if missing:
    print("Installing missing Python package(s): " + ", ".join(missing))
    subprocess.check_call(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            *missing,
        ]
    )
    importlib.invalidate_caches()
    print("Dependency installation completed.")
else:
    print("All required Python packages are available.")

import reference_dem_interface

importlib.reload(reference_dem_interface)

EXPECTED_VERSION = "1.0.1"

if reference_dem_interface.VERSION != EXPECTED_VERSION:
    raise RuntimeError(
        f"Expected reference_dem_interface v{EXPECTED_VERSION}, "
        f"but loaded v{reference_dem_interface.VERSION} from "
        f"{reference_dem_interface.__file__}."
    )

ui = reference_dem_interface.reference_dem_setup()
```
